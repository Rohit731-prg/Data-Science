```python
import pandas as pd

data = pd.read_csv("data.csv")
data.head(1)
```

```python
# Variance + Standard Deviation

# Variance ($\sigma^2$ or $s^2$): Measures spread by taking the average of the squared differences 
# from the mean. Because it squares everything, the units get squared too 
# (e.g., "squared dollars" 💲² or "squared test points" 💯²), which is hard to interpret!

# Standard Deviation ($\sigma$ or $s$): Is simply the square root of the variance.
# This brings the unit back to normal (e.g., dollars or test points), making it super easy to understand!

# 1. first calculate Variance:

variance = data['Data_value'].var()
variance
```

```python
# 2. Second Calculate Standard deviation

standard_deviation = data['Data_value'].std()
standard_deviation
```
