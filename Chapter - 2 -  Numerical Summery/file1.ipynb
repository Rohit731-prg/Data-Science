```python
# Quartiles 🍕 split your sorted dataset into four equal parts(25% of the data each)
# :Q1 (First Quartile / 25th Percentile): The middle number between the smallest value and the median.
# Q2 (Second Quartile / 50th Percentile): The median of the dataset.
# Q3 (Third Quartile / 75th Percentile): The middle number between the median and the largest value.

# IQR (Interquartile Range) 📏 is the distance between the third and first quartiles:
# IQR = Q3 - Q1

Q1 = data_value.quantile(0.25)
Q2 = data_value.quantile(0.50)
Q3 = data_value.quantile(0.75)

print(Q1, ":", Q2, ":", Q3)
```

```python
# The Interquartile Range (IQR)

IQR = Q3 - Q1
IQR
```
